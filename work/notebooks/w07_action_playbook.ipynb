```python
import os
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/ABK998/flyrank-ML/main/data/raw/content_refresh_anonymized.csv"
try:
    df = pd.read_csv(url)
except Exception:
    if os.path.exists("data/raw/content_refresh_anonymized.csv"):
        df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
    else:
        df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print(f"Loaded {len(df):,} rows for action playbook.")
```

## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
df["score"] = df["impressions_90d"].fillna(0) * (1 / (df["avg_position"].fillna(50)))
queue = df.sort_values(by="score", ascending=False).head(50).copy()
queue["action"] = np.where(queue["avg_position"].between(4, 15), "FULL_REFRESH", "PRIORITY_UPDATE")
queue["reason_code"] = np.where(queue["action"] == "FULL_REFRESH", "STRIKING_DISTANCE_DECLINE", "HIGH_TRAFFIC_STALE")
print("Operational review queue generated with action reason codes.")
```
