## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
feat_df = df.copy()
feat_df['impressions_90d'] = feat_df['impressions_90d'].fillna(0)
feat_df['content_age_days'] = feat_df['content_age_days'].fillna(feat_df['content_age_days'].median())
feat_df['avg_position'] = feat_df['avg_position'].fillna(100.0)
feat_df['ctr'] = feat_df['ctr'].fillna(0.0)
feat_df['word_count'] = feat_df['word_count'].fillna(feat_df['word_count'].median())

selected_features = ['impressions_90d', 'content_age_days', 'avg_position', 'ctr', 'word_count']
print(f'Successfully built feature matrix with shape: {feat_df[selected_features].shape}')
```
