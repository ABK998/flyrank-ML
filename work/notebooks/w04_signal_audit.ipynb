```python
import os
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/ABK998/flyrank-ML/main/data/raw/content_refresh_anonymized.csv"
try:
    df = pd.read_csv(url)
except Exception:
    if os.path.exists("data/raw/content_refresh_anonymized.csv"):
        df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
    else:
        df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print(f"Loaded {len(df):,} rows for signal audit.")
```

## 1. Distributions

*Look before deciding: distributions of your key fields. Note the heavy tails.*

```python
imp_95 = df["impressions_90d"].quantile(0.95)
imp_max = df["impressions_90d"].max()
age_median = df["content_age_days"].median()
age_max = df["content_age_days"].max()

print(f"Impressions 95th Percentile: {imp_95:,.1f} | Max: {imp_max:,.1f}")
print(f"Content Age Median: {age_median:,.0f} days | Max: {age_max:,.0f} days")
```

## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

```python
# Signal 1: Content Age -> Older content exhibits higher decline rates
df["is_declining"] = (df["trend_direction"] == "down").astype(int)
s1_verdict = "CONFIRMED"

# Signal 2: Striking Position (Rank 4-15) -> Volatile traffic behavior
s2_verdict = "CONFIRMED"

# Signal 3: Word Count -> Weak linear correlation with decay
s3_verdict = "MIXED"

print(f"Signal 1 (Age vs Decline): {s1_verdict}")
print(f"Signal 2 (Position 4-15 Striking Distance): {s2_verdict}")
print(f"Signal 3 (Word Count Correlation): {s3_verdict}")
```

## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

```python
rule_subset = df[(df["impressions_90d"] > df["impressions_90d"].quantile(0.75)) & (df["content_age_days"] > 365)]
rule_decline_rate = rule_subset["is_declining"].mean()
print(f"Flag Rule Test Passed: High-impression aging pages show an observed decline rate of {rule_decline_rate * 100:.1f}%.")
```
