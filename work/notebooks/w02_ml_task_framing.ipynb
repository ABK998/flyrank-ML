## 3. Success metric

Primary Metric: Precision@K (specifically Precision@50 and Precision@20).Formula:$$\text{Precision@K} = \frac{\text{True Positives in Top } K}{K}$$Why this metric: Content marketing and editorial teams operate under strict weekly capacity limits (e.g., refreshing 20 to 50 articles per cycle). High overall accuracy or ROC-AUC across 30,000 pages does not matter if the very first 50 pages recommended to an editor are false alarms. Precision@K evaluates model performance exactly at the point of decision-making.

## 5. Why ML beats a fixed rule here
* **Non-linear interactions:** Traffic decay cannot be captured by a simple rule like age > 180 days or impressions > 500. A page with high impressions but falling click-through rate, weak position, and poor scroll depth behaves very differently from an old evergreen page with low but stable traffic.
* **Empirical Evidence from Starter Pipeline:**
  * Fixed Heuristic Rule Precision@50: **0.240** (roughly 12 out of 50 right).
  * Machine Learning (Random Forest) Precision@50: **0.740** (roughly 37 out of 50 right).
  * The learned model achieves over a 3x improvement in ranking quality because it weighs complex, tangled features simultaneously rather than relying on arbitrary cutoff thresholds.
