## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Calculate the numbers
total_pages = len(df)
declining_pages = len(df[df["trend_direction"] == "down"])
decline_pct = (declining_pages / total_pages) * 100
median_impressions = df[df["trend_direction"] == "down"]["impressions_90d"].median()

print(f"Total pages: {total_pages}")
print(f"Actively declining pages: {declining_pages} ({decline_pct:.1f}%)")
print(f"Median 90-day impressions at stake per declining page: {median_impressions}")
```
